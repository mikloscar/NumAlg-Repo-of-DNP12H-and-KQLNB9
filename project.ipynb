```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(seed=42)
```

```python
#pseudo random code for 2D random walk with directions diagonal directions
rng = np.random.default_rng(seed=42)
steps = 1000

x = np.zeros(steps+1)
y = np.zeros(steps+1)

for i in range(1, steps+1):
    number = rng.integers(0, 4)
    if number == 0:
        x[i] = x[i-1] + 1
        y[i] = y[i-1]
    elif number == 1:
        x[i] = x[i-1] - 1
        y[i] = y[i-1]
    elif number == 2:
        x[i] = x[i-1]
        y[i] = y[i-1] + 1
    else:
        x[i] = x[i-1]
        y[i] = y[i-1] - 1



plt.plot(x, y, "-")
plt.scatter(x[0], y[0], label="Start")
plt.scatter(x[-1], y[-1], label="End")

plt.xlabel("x position")
plt.ylabel("y position")
plt.title("2D random walk")
plt.axis("equal")
plt.grid(True)
plt.legend()
plt.show()
```
